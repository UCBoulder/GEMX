```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import RegularGridInterpolator
from scipy.special import erfc

# Base directory containing scan folders
base_dir = "/pscratch/sd/h/haubrich/CST/runs/"

# Relevent Outputs
out_lambdadiv     = []
out_lambdamid     = []
out_lambdamid_w_e = []
out_udiv_avg      = []
out_lambda_avg    = []

# Loop over scan directories
for i in range(1, 260):  # adjust range as needed
    folder_name = f"scan_{i:04d}"   # zero-padded: 0001, 0002, ...
    # folder_name = f"psi_scan_{i:04d}"
    folder_path = os.path.join(base_dir, folder_name)

    # Check if directory exists
    if not os.path.isdir(folder_path):
        print(f"Skipping missing folder: {folder_name}")
        continue

    # print(f"Processing {folder_name}")

    ##################### Calculations ######################
    # =========================
    # Load data and define vars
    # =========================
    test_hit_div   = os.path.join(folder_path, "test_hit_div")
    test_hit_div_t = os.path.join(folder_path, "test_hit_div_t")
    initial_posi   = os.path.join(folder_path, "initial_posi")
    Rdat           = os.path.join(folder_path, "R.dat")
    Zdat           = os.path.join(folder_path, "Z.dat")
    psidat         = os.path.join(folder_path, "psi_p.dat")
    phidat         = os.path.join(folder_path, "phisolps.dat")

    p     = np.loadtxt(test_hit_div)
    hit_t = np.loadtxt(test_hit_div_t)
    ini_p = np.loadtxt(initial_posi)
    R     = np.loadtxt(Rdat)
    Z     = np.loadtxt(Zdat)
    psi   = np.loadtxt(psidat)
    phi   = np.loadtxt(phidat)
    psip  = psi.T
    phip  = phi.T
    
    # zero phi
    phip[:, :] = 0

    psi_a           = 0
    number_of_beams = 150
    dt              = 1500e-10
    mi              = 2 * 1.6726e-27
    e               = 1.6e-19
    T               = 200
    vth             = np.sqrt(T * e / mi)

    dR = abs(R[1]-R[0])
    dZ = abs(Z[1]-Z[0])
    
    # =========================
    # Divertor Plate Definition
    # =========================
    r1 = 1.2
    r2 = 1.5
    z1 = 0.25773 + Z[0]
    z2 = 0.25773 + Z[0]
    
    m  = (z2 - z1)/(r2 - r1)
    b  = z1 - m*r1
    
    R_div = np.linspace(r1, r2, 400)
    Z_div = np.linspace(z1, z2, 400)

    # =========================
    # Hit Calculation
    # =========================
    hit_num = np.zeros(int(np.max(hit_t)))

    for i in range(len(hit_t)):
        if hit_t[i] != 0:
            hit_num[int(hit_t[i]) - 1] += 1

    for i in range(1, len(hit_num)):
        hit_num[i] += hit_num[i - 1]

    # =========================
    # Midplane and Gradients
    # =========================
    zmid_id = np.argmin(np.abs(Z))
    Mid_z   = Z[zmid_id]
    Rmid_id = len(R) // 2

    sign_of_psi = (psip[-1,zmid_id]-psip[Rmid_id,zmid_id])/abs(psip[-1,zmid_id]-psip[Rmid_id,zmid_id])

    dpdR, dpdZ = np.gradient(psip, dR, dZ)
    gradpsiabs = np.sqrt(dpdR**2 + dpdZ**2)

    # =========================
    # X-point
    # =========================
    sub            = gradpsiabs[1:-2, 4:zmid_id-5]
    Z_x_id, R_x_id = np.unravel_index(np.argmin(sub), sub.shape)
    
    R_x_id += 1
    Z_x_id += 4
    
    R_x = R[R_x_id]
    Z_x = Z[Z_x_id]

    # =========================
    # Interpolators
    # =========================
    f    = RegularGridInterpolator((R, Z), psip)
    g    = RegularGridInterpolator((R, Z), gradpsiabs)
    phis = RegularGridInterpolator((R, Z), phip)

    # =========================
    # Find R_m_a
    # =========================
    for i in range(len(R)//2, len(R)):
        if (psip[i-1, zmid_id] - psi_a) * (psip[i, zmid_id] - psi_a) <= 0:    
            gradpsim = (gradpsiabs[i-1, zmid_id]*abs(psip[i, zmid_id] - psi_a) +
                        gradpsiabs[i, zmid_id]*abs(psip[i-1, zmid_id] - psi_a)) / \
                       abs(psip[i, zmid_id] - psip[i-1, zmid_id])
            idx_R_a = i
            break
    
    # Divertor intersection
    vals = np.array([f((R_div[i], Z_div[i])) for i in range(len(R_div))])
    idx_R_div = np.argmin(np.abs(vals))
    
    R_inter = R_div[idx_R_div]
    Z_inter = Z_div[idx_R_div]
    
    gradpsidiv = g((R_inter, Z_inter))

    # =========================
    # Particle loop
    # =========================
    Ep = []
    dpsi = []
    Rim = []
    udiv = []
    Rim_w_e = []
    
    number_of_particles = 0
    
    for i in range(len(p)):
        if p[i, 1] != 0:
            number_of_particles += 1
            dp = p[i, 0] - psi_a
            dpsi.append(dp)
    
            rim = sign_of_psi * dp / gradpsim
            Rim.append(rim)
    
            Rim_w_e.append(rim * p[i, 2])
            Ep.append(p[i, 2])
            udiv.append(abs(p[i, 1]))
    
    dpsi = np.array(dpsi)
    Rim = np.array(Rim)
    Rim_w_e = np.array(Rim_w_e)
    Ep = np.array(Ep)
    udiv = np.array(udiv)

    # =========================
    # Lambda calculations
    # =========================
    Ridiv = dpsi / gradpsidiv
    
    lambdadiv = np.sum(np.abs(Ridiv)) / len(dpsi)
    lambdamid = np.sum(np.abs(Rim)) / len(dpsi)
    lambdamid_w_e = np.sum(np.abs(Rim_w_e)) / np.sum(Ep)

    # =========================
    # Heat flux profile
    # =========================
    R_midgrid = np.linspace(np.min(Rim)-1e-4, np.max(Rim)+1e-4, number_of_beams+1)
    dRmid = (R_midgrid[-1] - R_midgrid[0]) / number_of_beams
    
    R_beam = 0.5 * (R_midgrid[:-1] + R_midgrid[1:])
    q_mid = np.zeros(len(R_beam))
    
    for i in range(len(dpsi)):
        pp = int(np.floor((Rim[i] - np.min(R_midgrid)) / dRmid))
        if 0 <= pp < len(q_mid):
            q_mid[pp] += Ep[i] / dRmid

    udiv_avg = np.sum(udiv) / len(udiv) / vth
    
    # =========================
    # Final calculations
    # =========================
    factor = 1
    
    lambda_avg = np.sum(q_mid * R_beam * factor) / np.sum(q_mid)
    
    # print("lambda_avg =", lambda_avg)

    out_lambdadiv.append(lambdadiv)
    out_lambdamid.append(lambdamid)
    out_lambdamid_w_e.append(lambdamid_w_e)
    out_udiv_avg.append(udiv_avg)
    out_lambda_avg.append(lambda_avg) 
    ###############################################################
# psi_scan = np.linspace(1,7,7)

# plt.figure()
# plt.plot(psi_scan*0.001,out_lambda_avg)
# plt.xlabel('$\\Delta \\psi$', fontsize=16)
# plt.ylabel('$\\lambda_q$', fontsize=16)
# plt.grid()
# plt.show()

# plt.figure()
# plt.hist(out_lambda_avg,20)
# plt.xlabel('$\\lambda_q$',fontsize=16)
# plt.ylabel('Frequency of Result',fontsize=16)
# # plt.grid()
# plt.show()

# Stack columns
data = np.column_stack((
    # out_lambdadiv,
    # out_lambdamid,
    # out_lambdamid_w_e,
    # out_udiv_avg,
    out_lambda_avg
))

# Save to CSV
np.savetxt(
    # "lambda_level_6_bug_fix.csv",
    "pseudo_random_lambda_outputs.csv",
    # "d_blob_scan.csv",
    np.array(out_lambda_avg)[:, None],
    delimiter=",",
    header="lambda_avg",
    comments=""
)

    
```
