```python
import numpy as np
import matplotlib.pyplot as plt

def read_data_file(filename):
    data   = []
    values = []
    with open(filename, 'r') as file:
        for line in file:
            values.append(line.strip().split())
        for i in range(0, len(values), 2):
            # Extract the two consecutive lines
            line1 = values[i]
            line2 = values[i + 1]
            # Create a tuple of four elements and append to the result list
            if len(line1) == 3 and len(line2) == 1:
                array_data = line1 + line2
                data.append(array_data)
    return np.array(data)

def parse_config_file(filepath):
    config = {}
    with open(filepath, 'r') as f:
        lines = f.readlines()
    i = 0
    while i < len(lines):
        if not lines[i].strip():
            i += 1
            continue
        label_line = lines[i].strip().strip("'").strip('"')
        i += 1
        if i >= len(lines):
            break
        value_line = lines[i].strip()
        i += 1
        labels = label_line.split()
        values = value_line.split()
        def try_convert(val):
            try:
                if '.' in val or 'e' in val.lower():
                    return float(val)
                else:
                    return int(val)
            except ValueError:
                return val
        converted_values = list(map(try_convert, values))
        if len(labels) == 1:
            config[labels[0]] = converted_values[0] if len(converted_values) == 1 else converted_values
        else:
            config.update(dict(zip(labels, converted_values)))
    return config

if __name__ == '__main__':
    filepath = 'gemx.in'
    parameters_dict = parse_config_file(filepath)

dt      = parameters_dict['dt']
omegaA  = 1539912.119669488  # Define omegaA (adjust the value as needed)
phirphi = np.loadtxt('testphi_r_phi')  # Load 'testphi_r_phi' data
phidiag     = np.loadtxt('testPhiDiag')
phidiag_fq  = np.loadtxt('testPhiFreq')
#ttt = np.loadtxt('testAtPhitrhotjt')  # Load 'testAtPhitrhotjt' data
# ttt         = read_data_file('testAtPhitrhotjt')
phirphi     = np.loadtxt('testphi_r_phi')  # Load 'testphi_r_phi' data
ex          = np.loadtxt('testER')
ez          = np.loadtxt('testEZ')
R           = np.loadtxt('R.dat')  # Load 'R.dat' data
Z           = np.loadtxt('Z.dat')  # Load 'Z.dat' data
dR          = (R[-1] - R[0]) / (len(R) - 1)
dZ          = (Z[-1] - Z[0]) / (len(Z) - 1)
BR          = np.loadtxt('BR.dat')  # Load 'BR.dat' data
BZ          = np.loadtxt('Bz.dat')  # Load 'Bz.dat' data
Bt          = np.loadtxt('Bt.dat')  # Load 'Bt.dat' data
mask2       = np.loadtxt('mask2')  # Load 'mask2' data
mask3       = np.loadtxt('mask3')  # Load 'mask3' data
mask4       = np.loadtxt('mask4')  # Load 'mask4' data
testphi     = np.loadtxt('testphi')  # Load 'testphi' data
testphiavg  = np.loadtxt('testphiavg') # Load 'testphiavg' data
# testout     = np.loadtxt('testoutput')
masktest    = np.loadtxt('mask')  # Load 'mask' data
#psitest = np.loadtxt('psi_test')  # Load 'psi_test' data
netest      = np.loadtxt('testne0')  # Load 'testne0' data
apar0       = np.loadtxt('testapar0')  # Load 'testapar0' data
ne2         = np.loadtxt('testne')  # Load 'testne' data
apar2       = np.loadtxt('testapar')  # Load 'testapar' data
jpar2       = np.loadtxt('testjpar')  # Load 'testjpar' data
jpar0       = np.loadtxt('testj0')  # Load 'testj0' data
ne0zeta     = np.loadtxt('testne0_zeta')  # Load 'testne0_zeta' data
j0zeta      = np.loadtxt('testjpar0_zeta')  # Load 'testjpar0_zeta' data
c2_over_vA2 = np.loadtxt('test_1_over_vA2')  # Load 'test_1_over_vA2' data

weights     = np.loadtxt('testweights')

fluxes      = np.loadtxt('test_fluxdiag')
# print(weights)

p = np.linspace(0, 2*np.pi, 258, endpoint=False)

# Initialize arrays to store loaded data (assuming the same dimensions as Matlab script)
phi2  = np.zeros((258, 258))
# tmp   = np.zeros((258, 258))
# out   = np.zeros((258,258))
aphi  = np.zeros((258, 258))
ER    = np.zeros((258,258))
EZ    = np.zeros((258,258))
a0    = np.zeros((258, 258))
m1    = np.zeros((258, 258))
psi   = np.zeros((258, 258))
n0    = np.zeros((258, 258))
n2    = np.zeros((258, 258))
a2    = np.zeros((258, 258))
j0    = np.zeros((258, 258))
j2    = np.zeros((258, 258))
m2    = np.zeros((258, 258))
m3    = np.zeros((258, 258))
m4    = np.zeros((258, 258))
vA2   = np.zeros((258, 258))
prphi = np.zeros((parameters_dict['kmx'], 258))

# Loop to populate arrays with loaded data
for i in range(258):
    for j in range(258):
        ii = ((i-1)*258 + j-1) // 3
        jj = (i-1)*258 + j - 3 * (((i-1)*258 + j-1) // 3) - 1 

        phi2[i, j] = testphi[ii, jj]
        # out[i,j]   = testout[ii,jj]
        aphi[i, j] = testphiavg[ii, jj]
        ER[i, j]   = ex[ii, jj]
        EZ[i, j]   = ez[ii, jj]
        a0[i, j] = apar0[ii, jj]
        m1[i, j] = masktest[ii, jj]
        #psi[i, j] = psitest[ii, jj]
        n0[i, j] = netest[ii, jj]
        n2[i, j] = ne2[ii, jj]
        a2[i, j] = apar2[ii, jj]
        j0[i, j] = jpar0[ii, jj]
        j2[i, j] = jpar2[ii, jj]
        m2[i, j] = mask2[ii, jj]
        m3[i, j] = mask3[ii, jj]
        m4[i, j] = mask4[ii, jj]
        vA2[i, j] = c2_over_vA2[ii, jj]

# Loop to populate prphi array
for i in range(parameters_dict['kmx']):
    for j in range(258):
        ii = ((i-1)*258 + j-1) // 3
        jj = (i-1)*258 + j - 3 * (((i-1)*258 + j-1) // 3) - 1
        prphi[i, j] = phirphi[ii, jj]


gamma = (1/2)*np.gradient(np.log(phidiag[:,1]),dt)
print(np.sum(gamma[800:])/len(gamma[800:])*10**(-4))


# Real Frequency Calculation
tmp_freq    = phidiag_fq[800:,1]
fft_phi     = np.fft.fft(tmp_freq)
freq        = np.fft.fftfreq(len(tmp_freq), d=dt)
freqindex   = np.arange(len(tmp_freq)//2)
fft_phi_mag = np.abs(fft_phi[freqindex])
freq_pos    = freq[freqindex]

peak_omega  = freq_pos[np.argmax(fft_phi_mag)]*2*np.pi
print(f"Dominant Omega_r: {peak_omega*10**(-4)/(2*np.pi)} /(10^4 rad/s)")


plt.figure()
# plt.countourfx,y)

phi_mesh = np.linspace(0, 2*np.pi, parameters_dict['kmx'])

r, theta = np.meshgrid(R, phi_mesh)

#-- Plot... ------------------------------------------------
fig, ax = plt.subplots(subplot_kw=dict(projection='polar'))
ax.contourf(theta, r, prphi,100)
ax.set_ylim(0,np.max(R))
plt.savefig("polar.png")
plt.show()
        
# Plotting figures
plt.figure()
plt.plot(weights[:,0],weights[:,1])
plt.xlabel('timestep')
plt.ylabel('$\\frac{\\sum w_p}{N}$')
# plt.savefig("weight_diag.png")
plt.grid()
plt.show()

# plt.figure()
# plt.plot(phidiag_fq[:,0],phidiag_fq[:,1])
# plt.xlabel('timestep')
# plt.ylabel('$\\phi$')
# plt.savefig("phisquare.png")
# plt.grid()
# plt.show()

plt.figure()
plt.plot(phidiag[:,0],phidiag[:,1])
plt.xlabel('timestep')
plt.ylabel('$|\\phi|^2$')
plt.savefig("phisquare.png")
plt.grid()
plt.show()

plt.figure()
plt.plot(phidiag[:,0],np.log(phidiag[:,1]))
plt.xlabel('timestep')
plt.ylabel('$ln(|\\phi|^2)$')
plt.savefig("phisquare.png")
plt.grid()
plt.show()

plt.figure()
plt.plot(phidiag[:,0],gamma)
plt.xlabel('timestep')
plt.ylabel('$\\gamma$')
plt.savefig("gamma.png")
plt.grid()
plt.show()

plt.figure()
plt.plot(fluxes[:,0],fluxes[:,1]/max(abs(fluxes[:,1])),label='Q')
plt.plot(fluxes[:,0],fluxes[:,2]/max(abs(fluxes[:,2])),label='$\\Gamma$')
# plt.plot(fluxes[:,0],fluxes[:,3]/max(fluxes[:,3]),label='$\\Pi$')
plt.legend()
plt.xlabel('timestep')
plt.ylabel('Norm. Flux')
plt.grid()
plt.savefig('fluxes.png')
plt.show()

plt.figure()
plt.contourf(R, Z, n2, 100)#,cmap='RdYlBu')
#plt.axis('equal')
plt.colorbar()
plt.xlabel('R(m)', fontsize=20)
plt.ylabel('Z(m)',fontsize=20)
plt.title(r'$n_{e}$', fontsize=24)
# plt.savefig('Figures/n_e.png')

# plt.figure()
# plt.contourf(R, Z, tmp, 100)
# #plt.axis('equal')
# plt.colorbar()
# plt.xlabel('R(m)', fontsize=20)
# plt.ylabel('Z(m)',fontsize=20)
# plt.title(r'$RHS$', fontsize=24)

plt.figure()
plt.contourf(R, Z, a2, 100)#,cmap='RdYlBu')
#plt.axis('equal')
plt.colorbar()
plt.xlabel('R(m)', fontsize=20)
plt.ylabel('Z(m)',fontsize=20)
plt.title(r'$\delta A_{||}$', fontsize=24)
# plt.savefig('Figures/delApar.png')

plt.figure()
plt.contourf(R, Z, j2, 100)#,cmap='RdYlBu')
#plt.axis('equal')
plt.colorbar()
plt.xlabel('R(m)', fontsize=20)
plt.ylabel('Z(m)',fontsize=20)
plt.title(r'$\mu_0 \delta j_{e||}$', fontsize=24)
# plt.savefig('Figures/mudelj.png')

plt.figure()
plt.contourf(R, Z, n0, 100)#,cmap='RdYlBu')
#plt.axis('equal')
plt.colorbar()
plt.xlabel('R(m)', fontsize=20)
plt.ylabel('Z(m)',fontsize=20)
plt.title(r'$n_{e0}$', fontsize=24)
# plt.savefig('Figures/n_e0.png')

plt.figure()
plt.contourf(R, Z, a0, 100)#,cmap='RdYlBu')
#plt.axis('equal')
plt.colorbar()
plt.xlabel('R(m)', fontsize=20)
plt.ylabel('Z(m)',fontsize=20)
plt.title(r'$A_{||0}$', fontsize=24)
# plt.savefig('Figures/Apar0.png')

plt.figure()
plt.contourf(R, Z, phi2, 100)#,cmap='RdYlBu')
plt.colorbar()
plt.contour(R, Z, m1 * 0.001, 1, colors='black', linewidths=2)
#plt.axis('equal')
#plt.colorbar()
plt.xlabel('R(m)', fontsize=20)
plt.ylabel('Z(m)',fontsize=20)
plt.title(r'$\phi_{end}$', fontsize=24)
# plt.savefig('Figures/phiend.png')

# plt.figure()
# plt.contourf(R, Z, out, 100)
# plt.colorbar()
# plt.contour(R, Z, m1 * 0.001, 1, colors='black', linewidths=2)
# plt.axis('equal')
# #plt.colorbar()
# plt.xlabel('R(m)', fontsize=20)
# plt.ylabel('Z(m)',fontsize=20)
# plt.title(r'out', fontsize=24)

plt.figure()
plt.contourf(R, Z, aphi, 100)#,cmap='RdYlBu')
plt.colorbar()
plt.contour(R,Z,m1 * 0.001, 1, colors='black', linewidths=2)
#plt.axis('equal')
plt.xlabel('R(m)', fontsize=20)
plt.ylabel('Z(m)', fontsize=20)
plt.title(r'$\left < \phi_{end} \right >$', fontsize=24)
plt.savefig('avgphiend.png')

plt.figure()
plt.contourf(R, Z, j0, 100)#,cmap='RdYlBu')
#plt.axis('equal')
plt.colorbar()
plt.xlabel('R(m)', fontsize=20)
plt.ylabel('Z(m)',fontsize=20)
plt.title(r'$j_{e||0}$', fontsize=24)
# plt.savefig('Figures/jpar0.png')

plt.figure()
plt.contourf(R, Z, vA2, 100)#,cmap='RdYlBu')
#plt.axis('equal')
plt.colorbar()
plt.xlabel('R(m)', fontsize=20)
plt.ylabel('Z(m)',fontsize=20)
plt.title(r'$1/v_A^2$', fontsize=24)
# plt.savefig('Figures/1_vA2.png')

# plt.figure()
# plt.plot(omegaA * dt * np.arange(len(ttt)), ttt[:, 0])
# plt.xlabel(r'$t \cdot \omega_A$')
# plt.ylabel(r'$\delta A_{||}$')
# plt.tick_params(axis='both', which='major', labelsize=24)

# plt.figure()
# plt.plot(omegaA * dt * np.arange(len(ttt)/2), ttt[:, 1])
# plt.xlabel(r'$t \cdot \omega_A$')
# plt.ylabel(r'$\delta \phi$')
# plt.tick_params(axis='both', which='major', labelsize=24)

# plt.figure()
# plt.plot(omegaA * dt * np.arange(len(ttt)/2), ttt[:, 2])
# plt.xlabel(r'$t \cdot \omega_A$')
# plt.ylabel(r'$e \mu_0 \delta n_e$')
# plt.tick_params(axis='both', which='major', labelsize=24)

# plt.figure()
# plt.plot(omegaA * dt * np.arange(len(ttt)/2), ttt[1::2, 0])
# plt.xlabel(r'$t \cdot \omega_A$')
# plt.ylabel(r'$\mu_0 \delta j_{e||}$')
# plt.tick_params(axis='both', which='major', labelsize=24)

# valuea = (m1 - 1) * a2
# valuej = (m2 - 2) * j2
# valuen = (m3 - 3) * n2
# valuephi = (m1 - 1) * phi2

# t = len(ttt) // 2
# tt = t * 2
# omega = np.arange(t) / t / omegaA

# plt.figure()
# plt.plot(ttt[::2, 1])
# ppp = np.fft.fft(ttt[::2, 1])
# pp = np.abs(ppp)
# for i in range(1, len(pp)):
#     pp[i] = 2 * pp[i] / t
# plt.figure()
# plt.plot(omega, pp)

#ER, EZ = np.gradient(-phi2, dR, dZ)

plt.figure()
plt.contourf(R, Z, ER, 100)#,cmap='RdYlBu')
plt.colorbar()
plt.xlabel('R(m)', fontsize=20)
plt.ylabel('Z(m)',fontsize=20)
plt.title(r'$E_R$', fontsize=24)
# plt.savefig('Figures/E_R.png')

plt.figure()
plt.contourf(R, Z, EZ, 100)#,cmap='RdYlBu')
plt.colorbar()
plt.xlabel('R(m)', fontsize=20)
plt.ylabel('Z(m)',fontsize=20)
plt.title(r'$E_Z$', fontsize=24)
# plt.savefig('Figures/E_Z.png')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import RegularGridInterpolator
# import imageio
# import cv2

# Constants
dt = 500e-10
steps = 200
omegaA = 1539912.119669488
filename = "testAnimated.gif"

# Load data
psi_p = np.loadtxt('psi_p.dat')  # 2D psi grid
phirphi = np.loadtxt('testphi_r_phi')  # Unused here
R = np.loadtxt('R.dat')
Z = np.loadtxt('Z.dat')
BR = np.loadtxt('BR.dat')
BZ = np.loadtxt('Bz.dat')
Bt = np.loadtxt('Bt.dat')
mask2 = np.loadtxt('mask2')
mask3 = np.loadtxt('mask3')
mask4 = np.loadtxt('mask4')
phit = np.loadtxt('testPhit')  # Large 2D array
mask = np.loadtxt('mask')      # Used to construct m1

# Construct m1 mask
m1 = np.zeros((258, 258))
for i in range(258):
    for j in range(258):
        ii = ((i * 258 + j) // 3) + 1
        jj = (i * 258 + j) - 3 * (((i * 258 + j) // 3))
        m1[i, j] = mask[ii-1, jj]

# Get psi contour data
fig = plt.figure()
psi_contours = plt.contour(R,Z,psi_p,40,levels=np.linspace(0,0.2,40))
plt.close(fig)

psi_values = psi_contours.levels

psi_paths = []
for collection in psi_contours.collections:
    for path in collection.get_paths():
        v = path.vertices  # shape (N, 2) --> [R, Z] pairs
        psi_paths.append(v)

# Create the GIF
images = []
for idx in range(steps):
    if (idx*100+100 > 1250):
        break
    pt = np.zeros((258, 258))
    for i in range(258):
        for j in range(258):
            step_data = phit[idx * 22188 : (idx + 1) * 22188, :]
            ii = ((i * 258 + j) // 3) + 1
            jj = (i * 258 + j) - 3 * (((i * 258 + j) // 3))
            if (idx + 1) * 22188 > phit.shape[0]:
                break
            pt[i, j] = phit[ii-1 + (idx * 22188), jj]
            idx_phit = ii-1 + (idx * 22188)
    
    # Create interpolator for this frame
    # pt_interp = RegularGridInterpolator((R, Z), pt.T)  # Note the transpose
    pt_interp = RegularGridInterpolator((R,Z), pt)

    # Sample pt along each psi_p contour path
    pt_along_psi = []
    for path in psi_paths:
        try:
            values = pt_interp(path)  # path = Nx2 array of [R, Z]
        except ValueError:
            # Some points might be out of bounds
            values = np.full(path.shape[0], np.nan)
        pt_along_psi.append(values)
    
    phi_contour_diag = []
    for contour in range(len(psi_paths)):
        # print(contour)
        phi_contour_diag.append(np.sum(pt_along_psi[contour]**2)/len(pt_along_psi[contour]))
        

    # --- Plot ---
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

    # Main 2D plot
    c1 = ax1.contourf(R, Z, pt, 20, cmap='viridis')
    ax1.contour(R, Z, m1 * 0.001, 1, colors='black', linewidths=2)
    ax1.set_title(f'Time Step {idx*100+100}')
    ax1.set_aspect('equal')
    fig.colorbar(c1, ax=ax1)

    # Average phi vs normalized psi
    ax2.plot(psi_values[1:]/np.max(psi_values), phi_contour_diag, color='blue')
    ax2.set_xlim(0, 1)
    ax2.set_xlabel('$\\psi_{Norm}$')
    ax2.set_ylabel('$|\\phi|^2$')
    # ax2.set_title('Flux-Surface Averaged ϕ')
    ax2.grid(True)

    # Save frame to image buffer
    fig.canvas.draw()
    frame = np.frombuffer(fig.canvas.tostring_rgb(), dtype=np.uint8)
    frame = frame.reshape(fig.canvas.get_width_height()[::-1] + (3,))
    images.append(frame)
    # plt.close(fig)

# Save GIF
imageio.mimsave(filename, images, format='GIF', duration=0.3)
```
